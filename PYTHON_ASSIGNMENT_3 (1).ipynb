```python
import random

# 1. Set Up the Game
secret_number = random.randint(1, 10)
attempts = 0
max_attempts = 3

print("Guess the number between 1 and 10")
print(f"You have {max_attempts} attempts")

# 2. + 3. Implement the Guessing Logic with while loop
while attempts < max_attempts:
    guess = int(input("Enter your guess: "))
    attempts += 1

    if guess < 1 or guess > 10:
        print("Invalid number. Must be between 1 and 10.")
        continue  # skip rest and go to next loop, but attempt is already counted

    if guess == secret_number:
        print(f"Correct! You guessed it in {attempts} attempts.")
        break  # exit loop because user won
    elif guess < secret_number:
        print("Too low! Try again.")
    else:  # guess > secret_number
        print("Too high! Try again.")

    # show remaining attempts
    print(f"Attempts left: {max_attempts - attempts}")

else:
    # This else runs only if while loop ends WITHOUT break
    # meaning user used all 3 attempts and didn't guess
    print(f"Game Over! You used all {max_attempts} attempts.")
    print(f"The correct number was {secret_number}")
```

```python
# 2)  FOR LOOP
# MULTIPLICATION TABLE GENERATOR

number = int(input("Enter a number for which you want to generate Multiplication table:"))

print(f"\nMultiplication table of {number} :")

for i in range(1,11):
  result = number * i
  print(f"{number} x {i} = {result}")
```
